## 1. Setup and Imports

```python
import pandas as pd
import numpy as np
import json
from pathlib import Path

print("Libraries loaded successfully.")
```

## 2. Load All Metric JSONs

```python
metrics_path = Path('../outputs/metrics')

metric_files = {
    'revenue_metrics': metrics_path / 'revenue_metrics.json',
    'channel_metrics': metrics_path / 'channel_metrics.json',
    'country_metrics': metrics_path / 'country_metrics.json',
    'category_metrics': metrics_path / 'category_metrics.json',
    'rfm_summary': metrics_path / 'rfm_summary.json',
    'anomaly_detection': metrics_path / 'anomaly_detection.json',
    'scenario_analysis': metrics_path / 'scenario_analysis.json',
    'recommendations': metrics_path / 'recommendations.json',
}

metrics = {}
for key, path in metric_files.items():
    if path.exists():
        with open(path, 'r') as f:
            metrics[key] = json.load(f)
        print(f"Loaded {key}")
    else:
        print(f"NOT FOUND: {key} at {path}")

loaded_count = len(metrics)
total_count = len(metric_files)
print(f"\nLoaded {loaded_count} of {total_count} metric files.")
print(f"Available: {list(metrics.keys())}")
```

## 3. Revenue Metrics

```python
if 'revenue_metrics' in metrics:
    print("REVENUE METRICS")
    print("=" * 60)
    rm = metrics['revenue_metrics']
    if isinstance(rm, dict):
        for key, value in rm.items():
            if isinstance(value, dict):
                print(f'\n{key}:')
                for k, v in value.items():
                    print(f'  {k}: {v}')
            else:
                print(f'  {key}: {value}')
    else:
        print(rm)
else:
    print("Revenue metrics not available.")
```

## 4. Channel Performance

```python
if 'channel_metrics' in metrics:
    print("CHANNEL METRICS")
    print("=" * 60)
    print(json.dumps(metrics['channel_metrics'], indent=2, default=str))
else:
    print("Channel metrics not available.")
```

## 10. Recommendations

```python
if 'recommendations' in metrics:
    print("KEY RECOMMENDATIONS")
    print("=" * 60)
    recs = metrics['recommendations']
    if isinstance(recs, list):
        for i, rec in enumerate(recs, 1):
            if isinstance(rec, dict):
                title = rec.get('title', f'Recommendation {i}')
                print(f'\n{i}. {title}')
                if 'description' in rec:
                    print(f'   {rec["description"]}')
                if 'priority' in rec:
                    print(f'   Priority: {rec["priority"]}')
                if 'impact' in rec:
                    print(f'   Expected Impact: {rec["impact"]}')
            else:
                print(f'\n{i}. {rec}')
    elif isinstance(recs, dict):
        print(json.dumps(recs, indent=2, default=str))
else:
    print("Recommendations not available.")
```

## 11. Executive Summary

```python
print('EXECUTIVE SUMMARY')
print("=" * 70)

if 'revenue_metrics' in metrics:
    rm = metrics['revenue_metrics']
    if isinstance(rm, dict):
        for key in ['total_revenue', 'total_orders', 'average_order_value']:
            if key in rm:
                print(f"  {key}: {rm[key]}")

if 'channel_metrics' in metrics:
    cm = metrics['channel_metrics']
    if isinstance(cm, dict) and 'top_channel' in cm:
        print(f"  Top channel: {cm['top_channel']}")

if 'country_metrics' in metrics:
    ctrm = metrics['country_metrics']
    if isinstance(ctrm, dict) and 'top_country' in ctrm:
        print(f"  Top country: {ctrm['top_country']}")

if 'category_metrics' in metrics:
    catm = metrics['category_metrics']
    if isinstance(catm, dict) and 'top_category' in catm:
        print(f"  Top category: {catm['top_category']}")

if 'anomaly_detection' in metrics:
    ad = metrics['anomaly_detection']
    if isinstance(ad, dict) and 'anomaly_count' in ad:
        print(f"  Anomalies detected: {ad['anomaly_count']}")

if 'recommendations' in metrics:
    recs = metrics['recommendations']
    if isinstance(recs, list):
        print(f"  Total recommendations: {len(recs)}")

print()
```
